```python
#위치 인수와 리스트 언패킹 사용하기

print(10,20,30)
```

```python
# 위치 인수를 사용하는 함수를 만들고 호출하기

def print_numbers(a,b,c):
    print(a)
    print(b)
    print(c)

print_numbers(10,20,30)
    
```

```python
# 언패킹 사용하기
x = [10,20,30]
print_numbers(x) # => ([10,20,30])

```

```python
# 언패킹 사용하기
x = [10,20,30]
print_numbers(*x) # => ([10,20,30])
```

```python
# 가변인수 함수 만들기

# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):
    for arg in args:
        print(arg)
```

```python
print_numbers(10)
```

```python
print_numbers(10,20,30,40)
```

```python
x = [10]
print_numbers(*x)
```

```python
y = [10,20,30,40]
print_numbers(*y)
```

```python
#키워드 인수 사용하기

def personal_info(name, age, address):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', address)
```

```python
personal_info('홍길동',30,'서울시 용산구 이촌동')
```

```python
#키워드 인수
#함수(키워드 = 값)

personal_info(name="홍길동",age = 30,address="서울시 용산구 이촌동")
```

```python
#키워드 인수와 딕셔너리 언패킹 사용하기

def personal_info(name, age, address):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', address)


    
```

```python
x = {
    'name': '홍길동', 
    'age': 30,
    'address': '서울시 용산구 이천동'
}
personal_info(**x)
```

```python
personal_info(**{
    'name': '홍길동', 
    'age': 30,
    'address': '서울시 용산구 이천동'
})
```

```python
# **를 두번 사용하는 이유 

x = {
    'name': '홍길동', 
    'age': 30,
    'address': '서울시 용산구 이천동'
}

personal_info(*x)
```

```python

x = {
    'name': '홍길동', 
    'age': 30,
    'address': '서울시 용산구 이천동'
}

personal_info(**x)
```

```python
# 키워드 인수를 사용한 가변 인수 함수만들기
#def 함수이름(**매개변수):
#   코드

def personal_info(**kwargs):
    for kw, arg in kwargs.items():
        print(kw, ':', arg, sep='')

x = {
    "name": '홍길동',
    "age": 30,
    "address": "서울시 용산구",
    "hobby": ["축구","농구","게임"]
}

personal_info(**x)
```

```python
personal_info(name= '홍길동')
```

```python
personal_info(name = '홍길동',age=30,address = '서울시 용산구 이천동')
```

```python
x = {'name':'홍길동'}
personal_info(**x)

y = {'name': '홍길동', 'age':30, 'address': '서울시 용산구 이천동'}
personal_info(**y)
```
