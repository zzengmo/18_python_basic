```python
#재귀호출 사용하기

def hello():
    print('hello world!')
    hello()

# hello()
```

```python
#재귀호출에 종료 조건 만들기

def hello(count):
    if count == 0:
        return

    print('Hello world!',count)
    count -= 1
    hello(count)

hello(5)
```

```python
#재귀호출로 팩토리얼 구하기

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
```
