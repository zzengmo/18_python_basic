```python
#3의 배수 일때 Fizz, 5의 배수일때 Buzz

for i in range(1,21):
    if i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1,21):
    if i % 3 == 0 and i % 5 == 0:
        print("fizzbuzz")
    elif i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 코드 단축하기

for i in range(1,31):
    print('Fizz' * (i % 3 == 0) + 'Buzz' * (i % 5 == 0) or i)
```
