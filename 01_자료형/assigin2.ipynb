```python
# 문제 14. 성적 관리 프로그램
# 학생들의 성적을 관리하는 프로그램을 작성하시오.

# 다음 학생들의 성적 정보를 딕셔너리로 저장하시오
# 학생1: 이름=김민수, 국어=85, 영어=90, 수학=78
# 학생2: 이름=이영희, 국어=92, 영어=88, 수학=95
# 학생3: 이름=박철수, 국어=78, 영어=85, 수학=82

# 요구사항:
# 1. 각 학생의 총점과 평균을 계산하여 딕셔너리에 추가
# 2. 각 학생의 정보를 보기 좋게 출력

students = {
    "김민수": {"국어":85, "영어":90, "수학":78},
    "이영희": {"국어":92, "영어":88, "수학":95},
    "박철수": {"국어":78, "영어":85, "수학":82},
}

김민수 = students["김민수"]
김민수["총점"] = 김민수["국어"] + 김민수["수학"] + 김민수["영어"]
김민수["평균"] = round(김민수["총점"] / 3, 2)

박철수 = students["박철수"]
박철수["총점"] = 박철수["국어"] + 박철수["수학"] + 박철수["영어"]
박철수["평균"] = round(박철수["총점"] / 3, 2)

이영희 = students["이영희"]
이영희["총점"] = 이영희["국어"] + 이영희["수학"] + 이영희["영어"]
이영희["평균"] = round(이영희["총점"] / 3, 2)

print(f"이름: 박철수")
print(f"국어: {박철수['국어']}점, 영어: {박철수['영어']}점, 수학: {박철수['수학']}점")
print(f"총점: {박철수['총점']}점, 평균: {박철수['평균']}점")

print(f"이름: 김민수")
print(f"국어: {김민수['국어']}점, 영어: {김민수['영어']}점, 수학: {김민수['수학']}점")
print(f"총점: {김민수['총점']}점, 평균: {김민수['평균']}점")

print(f"이름: 이영희")
print(f"국어: {이영희['국어']}점, 영어: {이영희['영어']}점, 수학: {이영희['수학']}점")
print(f"총점: {이영희['총점']}점, 평균: {이영희['평균']}점")
```

```python
# 문제 19. 두 반의 학생 명단 비교

# A반과 B반 학생 명단이 있습니다.


a_class = {'김철수', '이영희', '박민수', '최지영', '홍길동'}
b_class = {'박민수', '최지영', '정수민', '강호동', '이영희'}

# 다음 질문에 답하시오:


# 1. 두 반 모두에 있는 학생 (교집합)
a = a_class & b_class
print(a)
# 2. 두 반을 합친 전체 학생 (합집합)
b = a_class | b_class
print(b)
# 3. A반에만 있는 학생 (차집합)
c = a_class - b_class
print(c)

# 4. B반에만 있는 학생 (차집합)
d = b_class - a_class
print(d)
```
