```python
#람다 표현식과 map,filter,reduce 함수 활용하기

a = list(range(1,11))

# lambda x: str(x) if x % 3 == 0 else x
b = lambda x: str(x) if x % 3 == 0 else x
list(map(b,a))
```

```python
a = list(range(1,11))
list(map(lambda x: str(x) if x == 1 else float(x) if x == 2 else x + 10,a))
```

```python
def f(x):
    if x == 1:
        return str(x)
    elif x == 2:
        return float(x)
    else:
        return x + 10

a = list(range(1,11))
list(map(f,a))
```

```python
#map에 객체를 여러개넣기
a = list(range(1,6))
b = [2,4,6,8,10]

list(map(lambda x, y: x * y,a,b))
```

```python
#filter
#filter(함수,반복가능한객체)

def f(x):
    return x > 5 and x < 10

a = [8,3,2,10,15,7,1,9,0,11]
list(filter(f,a))
```

```python
a = [8,3,2,10,15,7,1,9,0,11]
list(filter(lambda x:x > 5 and x < 10, a))
```

```python
#reduce 사용하기
# import random
# radom.randit(1,2)
from functools import reduce
#reduce (함수,반복가능한객체)

def f(x,y):
    return x + y

a = [1,2,3,4,5]
reduce(f,a)
```

```python
a = [1,2,3,4,5]
from functools import reduce
reduce(lambda x,y: x + y,a)
```
