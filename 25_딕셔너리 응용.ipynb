```python
#반복문으로 딕셔너리의 키 값을 상을 모두 출력하기
x = {'a':10,'b':20,'c':30,'d':40}
for i in x :
    print(i, end=" ")
```

```python
# 딕셔너리의 키만 출력하기


for i  in x.keys():
    print(i)
```

```python
#딕셔너리의 값만 출력하기
for i in x.values():
    print(i)
```

```python
#딕셔너리 표현식 사용하기
keys = ['a','b','c','d']
x = {key: value for key,value in dict.fromkeys(keys).items()}
x
```

```python
keys = ["a","b","c","d"]
x = dict.fromkeys(keys)
x

{key:value for key,value in x.items()}
x
```

```python
{key: 0 for key in dict.fromkeys(["a", "b", "c","d"]).keys()}
```

```python
{value: 0 for value in {'a': 10,'b':20,'c':30,'d':40}.values()}
```

```python
#딕셔너리 표현식에 if 조건문 사용하기

x = {'a':10,'b':20,'c':30,'d':40}

for key, value in x.items():
    if value == 20:
        del x[key]

print(x)
```

```python
x = {'a':10,'b':20,'c':30,'d':40}

del x["b"]
print(x)
```

```python
x = {'a':10,'b':20,'c':30,'d':40}

x = {key: value for key, value in x.items() if value != 20}
x
```
