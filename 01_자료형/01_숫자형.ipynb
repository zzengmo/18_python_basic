```python
## 사칙 연산 
a = 3
b = 4

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
## 나눗셈 후 나머지를 변환하는 % 연산자
a = 7
b = 3

print(a % 3)
print(7 % 3)
```

```python
## 나눗셈 후 몫을 반환하는 // 연산자

a = 7
b = 4

print(a // b)
print(7/4)
print(7//4)
```
