```python
#14.7 심사문제 합격 여부 판단하기

k,e,m,s = map(int,input().split())

if not (0 <= k <= 100 and 0 <= e <= 100 and 0 <= m <= 100 and 0 <= s <= 100):
        print("잘못된 점수")

else:
    if (k + e + m + s) / 4 >= 80:
        print("합격")
    else:
        print("불합격")

```

```python
#15.4 심사문제 교통카드 시스템 만들기

balance = 9000
age = int(input())

if (7 < age <= 12):
    print(" 요금 650원입니다.\n","남은잔액",balance - 650,"원 남앗습니다")
elif (13 <= age <= 18):
    print(" 요금 1050원 입니다.\n","남은잔액",balance - 1050,"원 남앗습니다")
elif (19 <= age ):
    print(" 요금 1250원입니다.\n","남은잔액",balance - 1250,"원 남앗습니다")
else:
    print("무료입니다.")

```
