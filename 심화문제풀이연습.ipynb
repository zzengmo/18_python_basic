```python
# 표준 입력으로 정수 두개가 입력됩니다. 
# 다음소스코드를 완성하여 첫번째 정수와 두번째 정수 사이의 숫자 중 3으로 끝나지 않는 숫자가 출력되게 만드세요
# while True: 에 맞춰서 들여쓰기해주세요

# 테스트 케이스 예제: 1 20 >>> 1 2 4 5 6 7 8 9 10 11 12 14 15 16 17 18 19 20

# 힌트 코드
# start, stop = map(int,input().split())

# i = start

# while True:


#     print(i, end=' ')
#     i += 1

start, stop = map(int,input().split()) ##start,stop에 입력값을 정수형으로 받음
i = start ## i가 start이다

while True: ## while문 시작
    if i % 10 == 3: ## i를 10으로 나눈 나머지가 3이면
        i += 1 ## i를 1 증가시키고
        continue ## while문 처음으로 돌아감
    if i > stop: ## i가 stop보다 크면
        break ## while문 종료
    print(i, end=' ') ## i를 출력하고
    i += 1 ## i를 1 증가시킴
```

```python
#23.7 심사문제 지뢰찾기
##표준입력으로 2차원 리스트의 가로(col)와 세로(row)가 입력되고 그다음 줄 부터 리스트의 요소로 들어갈 문자가 입려됩니다.
##이때 2차원의 리스트 안에서는 * 지뢰이고 .은 지뢰가 아닙니다.
##지뢰가 아닌 요소에는 인접한 지뢰의 개수를 출력하는 프로그램을 만드세요
##여러 줄을 입력받으려면 다음과 같이 for 반복문에서 input을 호출한뒤 append로 각 줄을 추가하면됩니다.
##list안에 문자열을 넣으면 문자열이 문자 리스트로 변환됩니다.

col, row = map(int,input().split()) 
## col, row에 입력값을 정수형으로 받음 
##map함수는 여러개의 값을 받을때 사용하고 int는 정수형으로 바꿔주는 함수이고 input은 입력값을 받는 함수 split은 공백을 기준으로 나눠서 입력값을 받음

matrix = [] ## 2차원 리스트를 만들기위해 빈리스트를 만듬

for i in range(row): ## row만큼 반복문을 돌림
    matrix.append(list(input())) ## 각 줄마다 입력값을 받아서 list로 변환후 matrix에 추가

for i in range(row): ## row만큼 반복문을 돌림
    for j in range(col): ## col만큼 반복문을 돌림
        if matrix[i][j] == "*": ## 만약 matrix[i][j]가 *이면
            continue ## *이면 다음 반복문으로 넘어감
        count = 0 ## count라는 변수를 만들고 0으로 초기화
        for x in range(-1, 2): ## x라는 변수를 만들고 -1부터 1까지 반복문을 돌림
            for y in range(-1, 2): ## y라는 변수를 만들고 -1부터 1까지 반복문을 돌림
                ni = i + x ## ni라는 변수를 만들고 i에 x를 더함
                nj = j + y ## nj라는 변수를 만들고 j에 y를 더함
                if 0 <= ni < row and 0 <= nj < col: ## 만약 ni가 0이상 row미만이고 nj가 0이상 col미만이면
                    if matrix[ni][nj] == "*": ## 만약 matrix[ni][nj]가 *이면
                        count += 1 ## count를 1 증가시킴
        matrix[i][j] = str(count) ## matrix[i][j]에 count를 문자열로 변환해서 저장
    print("".join(matrix[i])) ## matrix[i]를 문자열로 변환해서 출력
```

```python
#26.9 심사문제 두숫자의 공약수를 세트 형태로 구하고 최종 결과는공약수의 합

a,b = list(map(int,input().split())) # a b라는 변수에 int정수값 여러개 함수를 입력받을수있게 split map함수 입력함수 input사용

a = set(i for i in range(1,a + 1) if a % i == 0) # a는  i라는 변수를 만들고 1 부터 a+1까지 반복을하고 만약에 a를 i로 나눴을때 나머지가 0이라면
b = set(i for i in range(1,b + 1) if b % i == 0) # b는  i라는 변수를 만들고 1 부터 a+1까지 반복을하고 만약에 b를 i로 나눴을때 나머지가 0이라면

divisor = a & b # disvisor라는 변수는 a 그리고 b and연산자
result = 0 #result 결과값은 0이다
if type(divisor) == set: ## 만약 divisor이라는 변수 타입이 set이라면 
    result = sum(divisor) ##result는 divisor값을 더한다

print(result) #result결과값을 출력해라
```

```python
#30.7 심사문제 가장 낮은 점수 높은 점수와 평균 점수를 구하는 함수 만들기

korean, english, mathematics,science = map(int,input().split()) ## k e m s 표준입력값 받는 코드

def get_min_max_score(*args): ##함수 get_min_max_score를 만들고 *args라는 매개변수를 받음
    return min(args), max(args) ## return는 min(args), max(args)를 반환한다

def get_average(**kwargs):  ##함수 get_average를 만들고 **kwargs라는 매개변수를 받음
    total = sum(kwargs.values()) ## total이라는 변수를 만들고 kwargs.values()를 합한값을 저장
    return total / len(kwargs) ## total을 kwargs의 길이로 나눈값을 반환 

min_score, max_score = get_min_max_score(korean,english,mathematics,science) 
## min_score, max_score라는 변수를 만들고 get_min_max_score함수에 korean,english,mathematics,science를 넣어서 반환값을 받음
average_score = get_average(korean=korean, english = english,
                            mathematics = mathematics, science = science)
## average_score라는 변수를 만들고 get_average함수에 korean=korean, english = english, mathematics = mathematics, science = science를 넣어서 반환값을 받음
print('낮은 점수: {0:.2f}, 높은 점수: {1:.2f}, 평균 점수: {2:.2f}'
        .format(min_score,max_score, average_score))
## format함수를 사용해서 min_score,max_score, average_score를 출력해라

min_score, max_score = get_min_max_score(english,science)
## min_score, max_score라는 변수를 만들고 get_min_max_score함수에 english,science를 넣어서 반환값을 받음
average_score = get_average(english=english, science=science)
## average_score라는 변수를 만들고 get_average함수에 english=english, science=science를 넣어서 반환값을 받음
print('낮은 점수: {0:.2f}, 높은 점수: {1:.2f}, 평균 점수: {2:.2f}'.format(min_score,max_score,average_score))
## format함수를 사용해서 min_score,max_score, average_score를 출력해라
```
