```python
# for 와 range 사용하기

# for 변수 in range(횟수):
    # 수행할코드

for i in range(10):
    print('hello world!')
```

```python
# 반복문에서 변수의 변화 확인

for i in range(10):
    print("hello world!", i)
```

```python
for number in range(5):
    print("hello world", number)
```

```python
#시작하는 숫자와 끝나는 숫자 지정하기
for a in range(5,12):
    print(a)
```

```python
# 증가폭 사용하기
for b in range(0,10,2):
    print(b)
```

```python
#숫자를 감소시키기

for c in range(10,0,-1):
    print(c)
```

```python
for i in reversed(range(10)):
    print("hello world!",i)
```

```python
#입력한 횟수대로 반복하기 

count = int(input())

for i in range(count):
    print("hello, world!" ,i)
```

```python
#시퀀스 객체로 반복하기

a = [10,20,30,40,50]
for i in a:
    print(i)
```

```python
fruits = ('apple','oranage','grape')
for fruit in fruits:
    print(fruit)
```

```python
for letter in "Python":
    print(letter, end=' ')
```
