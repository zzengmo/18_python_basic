```python
#집합 연산 사욜하기
##합 집합
#세트1 | 세트2
##set .union(세트1, 세트2) =>세트1.union(세트2)

a = {1,2,3,4}
b = {3,4,5,6}

print(a | b)
print(set.union(a,b))
```

```python
#교집합 
#세트1 & 세트2
# set.intersection(세트1,세트2)

print(a & b)
print(set.intersection(a,b))
```

```python
##차집합
## 세트1 - 세트2 or 세트2 - 세트1
##set.difference(세트1,세트2) 

print(a- b)
print(set.difference(a,b))
```

```python
## 대칭 차집합
#세트1 ^ 세트2
#set.symmetric_difference(세트1,세트2)

print(a ^ b)
print(set.symmetric_difference(a,b))
```

```python
#집합 연산후 할당 연산자 사용하기
# 세트1 |= 세트2
# 세트1.update(세트2)

a = {1,2,3,4}
a |= {5}
print(a)


a = {1,2,3,4}
a.update({5})

print(a)
```

```python
#intersection_update
#세트1 &= 세트2

a = {1,2,3,4}
a &= {0,1,2,3,4}
print(a)

a = {1,2,3,4}
a.intersection_update({0,1,2,3,4})
print(a)
```

```python
a = {1,2,3,4}
a ^= {3,4,5,6}
print(a)

a = {1,2,3,4}
a.symmetric_difference_update({3,4,5,6})
print(a)
```

```python
#반복문으로 세트의 요소를 모두 출력하기

# for 변수 in 세트:
#     반복할 코드

a = {1,2,3,4}

for i in a:
    print(i)
```

```python
#세트 표현식 사용하기
#{식 for in 객체}
#set(식 for 변수 in 객체)

a = {i for i in 'apple'}
a
```

```python
#세트 표현식에 if 조건문 사용하기
#{식 for 변수 in 세트 if 조건식}
#set(식 for 변수 in 세트 if 조건식)

a = {i for i in 'pineaplle' if i not in 'apl'}
print(a)
```
