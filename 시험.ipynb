```python
# input() 함수를 이용하여 URL을 입력받고, 프로토콜(Protocol), 도메인(Domain), 경로(Path)로 분리하여 출력하는 프로그램을 작성하세요. 
# (단, 입력되는 URL은 항상 프로토콜://도메인/경로 형식을 갖춘다고 가정합니다.)

# 입력값:
# https://www.example.com/login
# 출력값:
# 프로토콜: https
# 도메인: www.example.com
# 경로: login
# 요구사항:
# 1. split() 메서드를 활용하세요. 2. 입력함수부터 작성하세요.



url = input()

protocol, rest = url.split("://")
domain, path = rest.split("/", 1)

print(f"프로토콜: {protocol}")
print(f"도메인: {domain}")
print(f"경로: {path}") 
```

```python
# 표준 입력으로 첫 번째 줄에는 회원 정보의 키(문자열), 두 번째 줄에는 값(문자열 및 숫자)이 공백으로 구분되어 입력됩니다. 
# 이를 딕셔너리로 만들고, 키가 password인 항목을
# 삭제한 뒤 출력하는 프로그램을 작성하세요.


# 입력값:
# id password name age
# admin 1234 홍길동 30
# 출력값:
# {'id': 'admin', 'name': '홍길동', 'age': '30'}
# 요구사항:
# 1. dict(), zip() 함수를 사용하세요. 2. 반복문을 이용하세요
# 3. copy()를 이용하여 복사하여 반복문을 이용하세요. 4. 입력함수부터 작성하세요.

keys = input().split()
values = input().split()

user = dict(zip(keys, values))
user.pop("password", None)
print(user)
```

```python
# 다음은 도서관의 책 배치 정보를 담은 딕셔너리입니다. 요구사항에 맞게 데이터를 추출하는 코드를 작성하세요.

# library = {
#  "문학": [
#  {"제목": "1984", "저자": "조지 오웰", "재고": 3},
#  {"제목": "위대한 개츠비", "저자": "F. 스콧 피츠제럴드", "재고": 0}
#  ],
#  "과학": [
#  {"제목": "코스모스", "저자": "칼 세이건", "재고": 5},
#  {"제목": "이기적 유전자", "저자": "리처드 도킨스", "재고": 2}
#  ]
# }
# 요구사항:
# 1. 재고가 0인 책의 제목과 분야를 모두 출력하시오. 2. 전체 재고 수량의 합계를 모두 출력하세요

library = {
    "문학": [
        {"제목": "1984", "저자": "조지 오웰", "재고": 3},
        {"제목": "위대한 개츠비", "저자": "F. 스콧 피츠제럴드", "재고": 0}
    ],
    "과학": [
        {"제목": "코스모스", "저자": "칼 세이건", "재고": 5},
        {"제목": "이기적 유전자", "저자": "리처드 도킨스", "재고": 2}
    ]
}

for genre, books in library.items():
    for book in books:
        if book["재고"] == 0:
            print(f"제목: {book['제목']}, 분야: {genre}")

total_stock = sum(book["재고"] for books in library.values() for book in books)
print(f"전체 재고 수량의 합계: {total_stock}")
```

```python
# 입력값을 참조하여 첫 번째 정수부터 두 번째 정수까지 숫자를 출력합니다. 
# 단, 4의 배수일 때는 Four, 5의 배수일 때는 Five, 4와 5의 공배수(20의 배수)일 때는 FourFive를 출력하고 그 외에는 숫자를 그대로 출력하세요. 
# 입력값:
# 15 22
# 출력값:
# Five 16 17 18 19 FourFive 21 22
# 요구사항:
# 1. if ~ elif ~ else 문과 나머지 연산자 %를 사용하세요. 2. 출력값처럼 가로로 출력되게 하세요.

start, end = map(int, input().split())

result = []
for num in range(start, end + 1):
    if num % 20 == 0:
        result.append("FourFive")
    elif num % 4 == 0:
        result.append("Four")
    elif num % 5 == 0:
        result.append("Five")
    else:
        result.append(str(num))

print(' '.join(result))
```

```python
n = 4  # 마름모 크기 (홀수 권장)

# 위쪽 절반 (증가)
for i in range(n):
    print(' ' * (n - i - 1) + '*' * (2 * i + 1))

# 아래쪽 절반 (감소)
for i in range(n - 2, -1, -1):
    print(' ' * (n - i - 1) + '*' * (2 * i + 1))
```

```python
start, end = map(int, input().split())

result = []
for num in range(start, end + 1):
    if num % 2 != 0:
        result.append(num ** 3)

print(result)
```

```python
# 입력값 받기
salaries = list(map(int, input().split(',')))

# 내림차순 정렬
salaries.sort(reverse=True)

# 천 단위 콤마와 '원' 붙여서 출력
result = [f"{salary:,}원" for salary in salaries]

print(' '.join(result))
```

```python
# 입력값 받기 (9개의 숫자를 한 줄로 입력)
nums = list(map(int, input().split()))

# 3x3 2차원 리스트로 변환
matrix = [nums[0:3], nums[3:6], nums[6:9]]

# 전체 합에서 정중앙 요소를 뺀 값 = 테두리 요소들의 합
total = sum(matrix[i][j] for i in range(3) for j in range(3))
border_sum = total - matrix[1][1]

print(border_sum)
```

```python
def analyze_scores(*args):
    average = sum(args) / len(args)
    count = sum(1 for score in args if score >= average)
    print(f"평균: {average}")
    print(f"평균 이상인 수의 개수: {count}")

# 입력값 받기
scores = list(map(int, input().split()))

# 함수 호출 (가변 인수로 리스트를 풀어서 전달)
analyze_scores(*scores)
```

```python
inventory = {
    "노트북": 1500000,
    "마우스": 30000,
    "키보드": 50000,
    "모니터": 300000
}

def checkout(**kwargs):
    total = 0
    for item, count in kwargs.items():
        if item in inventory:
            price = inventory[item] * count
            total += price
            print(f"{item} {count}개: {price:,}원")
        else:
            print(f"판매하지 않는 상품입니다: {item}")
    print(f"총 결제 금액: {total:,}원")

checkout(노트북=1, 마우스=2)
print()  # 구분을 위한 빈 줄
checkout(모니터=2, 패드=1)
```
