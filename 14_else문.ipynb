```python
# #else 사용하기

# if 조건식:
#     수행할코드
# else:
#     수행할코드

x = 5

if x == 10:
    print("10입니다.")
else:
    print("10이 아닙니다.")
```

```python
#if 조건문의 동작 방식 알아보기
if True: 
    print("참")
else:
    print("거짓")

if False:
    print("참")
else:
    print("거짓")

if None:
    print("참")
else:
    print("거짓")
```

```python
# 조건문에 숫자 지정하기

if 0:
    print("참")
else:
    print("거짓")

if 1:
    print("참")
else:
    print("거짓")

if 0x1f:
    print("참")
else:
    print("거짓")

if 0b1000:
    print("참")
else:
    print("거짓")

if 13.5:
    print("참")
else:
    print("거짓")

```

```python
if 'hello':
    print('참')
else:
    print('거짓')

if '':
    print('참')
else:
    print('거짓')
```

```python
#조건식을 여러개 지정하기

x = 10
y = 20

if x == 10 and y == 20:
    print("참")
else:
    print("거짓")
```

```python
#중첩 if 조건문과 논리 연산자
x = int(input())

if x > 0:
    if x < 20:
        print("20보다 작은 양수입니다.")
```

```python
x = 10

if x > 0 and x < 20:
    print("20보다 작은 양수입니다.")
```
