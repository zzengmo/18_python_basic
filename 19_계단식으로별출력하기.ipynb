```python
# 중첩 루프 사용하기

for i in range(5):
    for j in range(5):
        if j <= i:
            print('*', end='')
    print()
```

```python
for i in range(5):
    for j in range(5):
        print("j:", j, sep="", end=" ")
    print("i:",i,"\\n",sep="")
```

```python
for i in range(5): ## for문 0부터4까지 5번 반복 세로방향
    for j in range(5): ## for문 0부터4까지 5번 반복 가로방향
        if j <= i: ## 만약 j가 i보다 작거나 같으면
            print('*',end='')  ## 별을 출력하고 줄바꿈하지 않음
    print() ## 줄바꿈을 위해 print()를 사용
```

```python
for i in range(5): 
    for j in range(5):
        print("*", end='')
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j >= i:
            print('*',end='')
    print()
```

```python
for i in range(5):
    print('*')
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*",end="")
        else:
            print(" ",end="")
    print()
```
