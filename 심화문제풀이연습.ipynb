```python
# 표준 입력으로 정수 두개가 입력됩니다. 
# 다음소스코드를 완성하여 첫번째 정수와 두번째 정수 사이의 숫자 중 3으로 끝나지 않는 숫자가 출력되게 만드세요
# while True: 에 맞춰서 들여쓰기해주세요

# 테스트 케이스 예제: 1 20 >>> 1 2 4 5 6 7 8 9 10 11 12 14 15 16 17 18 19 20

# 힌트 코드
# start, stop = map(int,input().split())

# i = start

# while True:


#     print(i, end=' ')
#     i += 1

start, stop = map(int,input().split()) ##start,stop에 입력값을 정수형으로 받음
i = start ## i가 start이다

while True: ## while문 시작
    if i % 10 == 3: ## i를 10으로 나눈 나머지가 3이면
        i += 1 ## i를 1 증가시키고
        continue ## while문 처음으로 돌아감
    if i > stop: ## i가 stop보다 크면
        break ## while문 종료
    print(i, end=' ') ## i를 출력하고
    i += 1 ## i를 1 증가시킴
```

```python
## 표준 입력으로 금액(정수)이 입력됩니다.
## 1회당 요금은 1350원으로 교통카드를 사용할 때마다 각줄에 잔액을 출력하는 프로그램을 만드세요
##최초 금액은 출력하지 않아야합니다.
##그리고 잔액은 음수가 될 수 없으며 잔액이 부족하면 출력을 끝냅니다.

a = int(input()) ## a는 내가 가지고있는 금액 정수 입력값 int 그리고 입력함수 input사용
b = 1350 ##b는 요금 1350원

while a >= b: ## while문 내가 가지고있는돈이 1회 요금보다 크거나 같다면
    a -= b  ## a = a - b ## 내가 가진돈에서 1350원을 뺌
    print(a) ##출력을 순서대로 나오게됨
```

```python
## 표준입력으로 정수가 입력됩니다.
## 입력된 정수의 구구단을 출력하는 프로그램을 만드세요
## 출력형식은 숫자 * 숫자 = 수자 처럼 만들고 숫자와 * , = 사이에는 공백을 한칸 뛰웁니다.

a = int(input()) ## a = 입력받는 값 ## 정수 입력이기때문에 int사용 , 입력함수 input사용

for i in range(1,10): ##for문 i는 1부터 9까지 반복
    print(a,"*",i,"=",a * i) ## a는 입력값 i는 1~9까지 그래서 내가 2를 입력하면 2*1 2*2 2*3 ... 쭈루룩 나옴
```

```python
##표준 입력으로 나이가 입력됩니다. 
## 교통카드 시스템에서 시내버스 요금은 다음과 같으며 각 나이에 맞게 요금을 차감한 뒤 잔액이 출력되게 만드세요
## if,elif 사용 
## 현재 교통카드에는 9000원 들어있음

#어린이 만7세이상 12세이하 650원
#청소년 만13세이상 18세이하 1050원
#성인 만19세이상 1250원

age = int(input()) ##표준입력 정수값int사용 입력함수 input사용
cash = 9000 ## 잔액이 9000원 있음

if 7 <= age <= 12: ##입력값 age가 7이상 12이하면 
    print(cash - 650)## 잔액cash에서 650원을 빼라
elif 13 <= age <= 18: ## 입력값 age가 13이상 18이하면
    print(cash - 1050) ## 잔액 cash에서 1050원을 빼라
elif age >= 19: ## 입력값 age가 19세이상이면 
    print(cash - 1250) ## 잔액cash에서 1250을 빼라
else: ##그외에 age입력값이 나오면 
    print("무료입니다.") ##문제에서는 아무말이 없지만 현실에서는 무료이닌깐 무료입니다.
```

```python
##표준 입력으로 국어 영어 수학 과학 점수가 입력됩니다.
##여기서 네 과목의 평균 점수가 80점이상 일때 합격
##평균점수에따라 합격 불합격 여부를 출력하는 프로그램을 만드세요
##만약 입력값이 100을 초과하면 잘못된 점수라고 나오게 출력하시오

k,e,m,s = map(int,input().split()) 
##변수 k e m s 값을 4개를 받아야하기때문에 map함수 그리고 정수값 받아야하기때문에 int 입력함수input을 사용 split()으로 공백을 기준으로 나눠서 입력값을 받음

if not (0 <= k <= 100 and 0 <= e <= 100 and 0 <= m <= 100 and 0 <= s <= 100): ## 만약 전부다 100점사이가 아니라면 
    print("잘못된 점수 입력") ##잘못된 점수 입력이라고 출력
else: ##그외에 전부다 100점사이이면
    if (k + e + m + s) / 4 >= 80: ##입력받은 4개의 합 나누기4 평균을 구하고 이게 80점이상이라면
        print("합격") ## 합격 출력
    else: ##아니라면
        print("불합격") ##불합격 출력
```

```python
## 표준 입력으로 가격과 쿠폰 이름이 각 줄에 입력됩니다.
## Cash3000 쿠폰은 3000원, Cahs5000쿠폰은 5000원 할인합니다.
## 쿠폰에 따라 할인된 가격을 출력하는 프로그램을 만드시오

x = int(input()) ## 입력값이 정수 int 사용 입력함수 input사용
coupon = input() ## 어떤 쿠폰을 사용할지 문자값 입력함수 input사용

if coupon == "Cash3000": ## 만약 쿠폰입력값에 Cash3000쿠폰이 입력이됬다면
    print(x - 3000) ## x 정수값에서 3000을 빼라

if coupon == "Cash5000": ## 만약 쿠폰입력값에 Cash5000쿠폰이 입력이됬다면
    print(x - 5000)  ## x 정수값에서 5000을 빼라


```

```python
# 표준 입력으로 정수 두개가 입력됩니다. 
# 다음소스코드를 완성하여 첫번째 정수와 두번째 정수 사이의 숫자 중 3으로 끝나지 않는 숫자가 출력되게 만드세요
# while True: 에 맞춰서 들여쓰기해주세요

# 테스트 케이스 예제: 1 20 >>> 1 2 4 5 6 7 8 9 10 11 12 14 15 16 17 18 19 20

# 힌트 코드
# start, stop = map(int,input().split())

# i = start

# while True:



#     print(i, end=' ')
#     i += 1

start, stop = map(int,input().split())

i = start

while True:
    if i % 10 == 3:
        i += 1
        continue
    if i > stop:
        break
    print(i, end=' ')
    i += 1
```

```python
##표준입력으로 국어 영어 수학 과학 점수가 입력됩니다.
## 평균 점수를 출력하는 프로그램을 만드세요 (정수로 출력)

k,e,m,s = map(int,input().split()) 
## 국어 영어 수학 과학의 변수를 만들고 map함수로 입력값을 정수형으로 받음 int는 정수형으로 바꿔주는 함수이고
## input은 입력값을 받는 함수 split은 공백을 기준으로 나눠서 입력값을 받음
avg = (k + e + m + s) / 4 ### 평균을 구하기 위해서 4과목의 합을 4로 나누기

print(int(avg)) ## 평균값을 정수형으로 바꿔서 출력
```

```python
## 표준입력으로 년 월 일 시 분 초 가 입력됩니다. 다음소스코드를 완성하여 입력된 날짜를 
## 년-월-일T시:분:초 형식으로 출력하게 만드시오

year,month,day,hour,minute,second = input().split() ##년 월 일 시 분 초 라는 변수를 한칸공백으로 입력값을 받을수있게 split와 입력함수input을 사용
print(year,month,day, sep="-",end="T") ##   year month day 변수에 sep를 사용해서 공백에 -을 넣어서 출력을하고 end를 사용해서 T를 넣어서 출력을해라
print(hour,minute,second, sep=":") ##   hour minute second 변수에 sep를 사용해서 공백에:을 넣어서 출력을해라
```
