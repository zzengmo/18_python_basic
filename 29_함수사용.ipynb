```python
#덧셈함수만들기
#def 함수이름(매개변수1, 매개변수2):
    # 코드

def add(a,b):
    #a = 10, b = 20 
    print(a+b)

result1 = add(10,20)
print(result1)
```

```python
#함수의 결과를 반환하기
#def 함수이름(매개변수):
    # return반환값

def add(a,b):
    return a + b

result2 = add(10,20)
print(result2)
```

```python
#1. 매개변수 없고 , 리턴값 없음 

def say_hello():
    print("Hello")
    
say_hello()
```

```python
#2. 매개변수 없고 리턴값은 있는 함수

def one():
    return 1

one()
```

```python
#3 매개변수 있고 리턴값은 없음

def add(a,b):
    print(a+b)
    
```

```python
#4 매개변수있고 리턴값도 있음

def add(a,b):
    return a + b

add(10,20)
```

```python
#return 으로 함수 중간에서 빠져나오기

def not_ten(a):
    if a == 10:
        return
    print(a,"입니다.", sep="")
```

```python
not_ten(5)
```

```python
not_ten(10)
```

```python
# 함수에서 값을 여러개 반환하기
def add_sub(a,b):
    return a + b, a - b

result = add_sub(10,20)
result
```

```python
x, y = add_sub(10,20)
print(x)
print(y)
```

```python
#함수의 호출 과정 알아보기

def mul(a,b):
    c = a * b
    return c

def add(a,b):
    c = a + b
    print(c)
    d = mul(a,b)
    print(d)

x = 10
y = 20
add(x,y) 
```
