```python
# 문자열 판단하기

import re

print(re.match('hello','hello world!'))
print(re.match('python','hello world!'))
```

```python
#문자열이 맨 앞에 오는지 맨뒤에 오는지 판단하기 
#^문자열
#문자열$
#re.search("패턴", "문자열")

re.search("^Hello", "Hello World!")
re.search("World!$","Hello World!")
```

```python
#지정된 문자열이 하나라도 포함되는지 판단하기

re.match("hello|word", "hello")
```

```python
# 범위 판단하기
print(re.match("[0-9]*",'1234'))
print(re.match("[0-9]+",'1234'))
print(re.match("[0-9]+",'abcd'))
print(re.match("a*b",'b'))
print(re.match("a+b",'b'))
print(re.match("a*b",'aab'))
print(re.match("a+b",'aab'))
```

```python
#문자가 한 개만 있는지 판단하기

re.match('abc?d','abd')
re.match('abc[0-9]?d','ab3c')
re.match('ab.d','abxd')
```

```python
#문자 개수 판단하기

print(re.match('h{3}','hhhello'))
print(re.match('(hello){3}','hellohellohelloworld'))
```

```python
re.match('[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}','02-100-1000')
```

```python
#숫자와 영문 문자를 조합해서 판단하기
re.match('[a-zA-Z0-9]+','Hello1234')
```

```python
re.match('[가-힣]+','홍길동')
```

```python
#특정문자 범위에 포함되지 않는지 판단하기
# [^범위]
# [^범위]+

print(re.match('[^A-Z]+','Hello'))
print(re.match('[^A-Z]+','hello'))
```

```python
# ^[범위]
# ^[범위]+

print(re.search('^[A-Z]+','Hello'))
```
